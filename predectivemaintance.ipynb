```python
df['Temp_Difference'] = df['Process temperature [K]'] - df['Air temperature [K]']

df['Power_Index'] = df['Rotational speed [rpm]'] * df['Torque [Nm]']

df['Torque_RPM_Ratio'] = df['Torque [Nm]'] / df['Rotational speed [rpm]']
```

```python
# Dependent Variable (Target)
y = df[['Machine failure']]

# Independent Variables (Features)
x = df.drop('Machine failure', axis=1)

print("Independent columns:", x.shape)
print("Dependent columns:", y.shape)
```

```python
print("Independent Variables:")
print(x.columns)

print("\nDependent Variable:")
print(y.columns[0] if hasattr(y, "columns") else y.name)
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, random_state = 42, test_size = 0.2)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

# Model Create
model = RandomForestClassifier(random_state=42)

# Train Model
model.fit(x_train, y_train)

# Prediction
y_pred = model.predict(x_test)

# Accuracy
accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
print("Accuracy (%):", accuracy * 100)
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score

print("Precision:", precision_score(y_test, y_pred))
print("Recall:", recall_score(y_test, y_pred))
print("F1 Score:", f1_score(y_test, y_pred))
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy*100:.2f}%")
```

```python
from sklearn.metrics import confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(5,4))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix")
plt.show()
```

```python
import pandas as pd
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# y_test agar DataFrame hai to Series bana do
y_test_series = y_test.squeeze()

comparison = pd.DataFrame({
    "Algorithm": [
        "Random Forest",
        "XGBoost",
        "Gradient Boosting"
    ],
    "Accuracy": [
        accuracy_score(y_test_series, y_pred),
        accuracy_score(y_test_series, y_pred_xgb),
        accuracy_score(y_test_series, y_pred_gb)
    ],
    "Precision": [
        precision_score(y_test_series, y_pred),
        precision_score(y_test_series, y_pred_xgb),
        precision_score(y_test_series, y_pred_gb)
    ],
    "Recall": [
        recall_score(y_test_series, y_pred),
        precision_score(y_test_series, y_pred_xgb),
        recall_score(y_test_series, y_pred_gb)
    ],
    "F1 Score": [
        f1_score(y_test_series, y_pred),
        f1_score(y_test_series, y_pred_xgb),
        f1_score(y_test_series, y_pred_gb)
    ]
})

comparison = comparison.round(4)
comparison
```

```python
best_model = comparison.sort_values(by="Accuracy", ascending=False)

print(best_model)
```
